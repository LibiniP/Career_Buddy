```python
"""
Test Suite for Goal-Based Scheduling Edge Cases
Tests various scenarios to ensure robust scheduling
"""

import sys
from pathlib import Path
BASE_DIR = Path.cwd()   # Jupyter notebook's working directory
sys.path.insert(0, str(BASE_DIR))

from agents.modules.goal_based import GoalBasedModule
from config.career_database import CAREER_DATABASE


def test_edge_case_1_high_availability():
    """
    Test Case 1: User has very high daily time availability (8 hours)
    Expected: All tasks should be scheduled, possibly in fewer days
    """
    print("\n" + "="*70)
    print("TEST 1: High Time Availability (8 hours/day)")
    print("="*70)
    
    # Create mock careers
    careers = [
        {
            'career': {
                'domain': 'Content Writer',
                'modules': [
                    {'name': 'SEO fundamentals', 'duration': 60},
                    {'name': 'Write articles', 'duration': 90},
                    {'name': 'Create portfolio', 'duration': 120}
                ]
            }
        },
        {
            'career': {
                'domain': 'Graphic Designer',
                'modules': [
                    {'name': 'Design principles', 'duration': 90},
                    {'name': 'Learn tools', 'duration': 120},
                    {'name': 'Create portfolio', 'duration': 150}
                ]
            }
        }
    ]
    
    module = GoalBasedModule(daily_hours=8.0)
    schedule = module.generate_daily_schedule(careers)
    
    print(f"✓ Schedule generated: {len(schedule)} days")
    
    # Verify all tasks are scheduled
    total_tasks = sum(len(career['career']['modules']) for career in careers)
    scheduled_tasks = sum(len(day['tasks']) for day in schedule)
    
    assert scheduled_tasks == total_tasks, f"Missing tasks! Expected {total_tasks}, got {scheduled_tasks}"
    print(f"✓ All {total_tasks} tasks scheduled correctly")
    
    module.print_schedule()
    
    return len(schedule) > 0


def test_edge_case_2_low_availability():
    """
    Test Case 2: User has very low daily time availability (1 hour)
    Expected: Tasks spread across more days, but all should be scheduled
    """
    print("\n" + "="*70)
    print("TEST 2: Low Time Availability (1 hour/day)")
    print("="*70)
    
    careers = [
        {
            'career': {
                'domain': 'Content Writer',
                'modules': [
                    {'name': 'SEO fundamentals', 'duration': 60},
                    {'name': 'Write articles', 'duration': 90},
                    {'name': 'Create portfolio', 'duration': 120}
                ]
            }
        }
    ]
    
    module = GoalBasedModule(daily_hours=1.0)
    schedule = module.generate_daily_schedule(careers)
    
    print(f"✓ Schedule generated: {len(schedule)} days")
    
    # Verify all tasks are scheduled
    total_tasks = sum(len(career['career']['modules']) for career in careers)
    scheduled_tasks = sum(len(day['tasks']) for day in schedule)
    
    assert scheduled_tasks == total_tasks, f"Missing tasks! Expected {total_tasks}, got {scheduled_tasks}"
    print(f"✓ All {total_tasks} tasks scheduled correctly")
    
    # Check that tasks exceeding daily capacity are still scheduled
    for day in schedule:
        print(f"  Day {day['day']}: {len(day['tasks'])} task(s), {day['total_time']} mins")
    
    module.print_schedule()
    
    return len(schedule) > 0


def test_edge_case_3_single_career():
    """
    Test Case 3: User selects only one career
    Expected: All modules from that career should be scheduled
    """
    print("\n" + "="*70)
    print("TEST 3: Single Career Selection")
    print("="*70)
    
    careers = [
        {
            'career': {
                'domain': 'Graphic Designer',
                'modules': [
                    {'name': 'Design principles', 'duration': 90},
                    {'name': 'Color theory', 'duration': 60},
                    {'name': 'Typography', 'duration': 45},
                    {'name': 'Learn tools', 'duration': 120},
                    {'name': 'Create portfolio', 'duration': 150}
                ]
            }
        }
    ]
    
    module = GoalBasedModule(daily_hours=3.0)
    schedule = module.generate_daily_schedule(careers)
    
    print(f"✓ Schedule generated: {len(schedule)} days")
    
    # Verify all tasks are scheduled
    total_tasks = len(careers[0]['career']['modules'])
    scheduled_tasks = sum(len(day['tasks']) for day in schedule)
    
    assert scheduled_tasks == total_tasks, f"Missing tasks! Expected {total_tasks}, got {scheduled_tasks}"
    print(f"✓ All {total_tasks} tasks scheduled correctly")
    
    module.print_schedule()
    
    return len(schedule) > 0


def test_edge_case_4_many_careers():
    """
    Test Case 4: User selects many careers (5+)
    Expected: Tasks should be interleaved for variety
    """
    print("\n" + "="*70)
    print("TEST 4: Multiple Career Selection (5 careers)")
    print("="*70)
    
    careers = []
    for key in ['teaching', 'cooking', 'writing', 'design', 'fitness']:
        if key in CAREER_DATABASE:
            careers.append({
                'career': CAREER_DATABASE[key]
            })
    
    module = GoalBasedModule(daily_hours=4.0)
    schedule = module.generate_daily_schedule(careers)
    
    print(f"✓ Schedule generated: {len(schedule)} days")
    
    # Verify all tasks are scheduled
    total_tasks = sum(len(career['career']['modules']) for career in careers)
    scheduled_tasks = sum(len(day['tasks']) for day in schedule)
    
    assert scheduled_tasks == total_tasks, f"Missing tasks! Expected {total_tasks}, got {scheduled_tasks}"
    print(f"✓ All {total_tasks} tasks scheduled correctly")
    
    # Check for career variety in early days
    if len(schedule) > 0:
        day1_careers = set(task['career'] for task in schedule[0]['tasks'])
        print(f"✓ Day 1 includes tasks from {len(day1_careers)} different careers (variety)")
    
    module.print_schedule()
    
    return len(schedule) > 0


def test_edge_case_5_empty_input():
    """
    Test Case 5: Empty or invalid input
    Expected: Should handle gracefully without crashing
    """
    print("\n" + "="*70)
    print("TEST 5: Empty/Invalid Input Handling")
    print("="*70)
    
    module = GoalBasedModule(daily_hours=2.0)
    
    # Test empty career list
    schedule1 = module.generate_daily_schedule([])
    assert schedule1 == [], "Empty input should return empty schedule"
    print("✓ Empty career list handled correctly")
    
    # Test career with no modules
    schedule2 = module.generate_daily_schedule([
        {'career': {'domain': 'Test', 'modules': []}}
    ])
    assert len(schedule2) == 0, "Career with no modules should return empty schedule"
    print("✓ Career with no modules handled correctly")
    
    return True


def test_edge_case_6_large_task():
    """
    Test Case 6: Task duration exceeds daily availability
    Expected: Task should still be scheduled (alone on a day)
    """
    print("\n" + "="*70)
    print("TEST 6: Task Larger Than Daily Capacity")
    print("="*70)
    
    careers = [
        {
            'career': {
                'domain': 'Content Creator',
                'modules': [
                    {'name': 'Small task', 'duration': 30},
                    {'name': 'HUGE task', 'duration': 300},  # 5 hours
                    {'name': 'Normal task', 'duration': 60}
                ]
            }
        }
    ]
    
    module = GoalBasedModule(daily_hours=2.0)  # Only 2 hours available
    schedule = module.generate_daily_schedule(careers)
    
    print(f"✓ Schedule generated: {len(schedule)} days")
    
    # Verify all tasks are scheduled, including the large one
    total_tasks = 3
    scheduled_tasks = sum(len(day['tasks']) for day in schedule)
    
    assert scheduled_tasks == total_tasks, f"Missing tasks! Expected {total_tasks}, got {scheduled_tasks}"
    print(f"✓ All {total_tasks} tasks scheduled, including large task")
    
    # Find the day with the large task
    large_task_day = None
    for day in schedule:
        for task in day['tasks']:
            if task['duration'] == 300:
                large_task_day = day
                break
    
    assert large_task_day is not None, "Large task not found in schedule"
    print(f"✓ Large task scheduled on day {large_task_day['day']}")
    
    module.print_schedule()
    
    return len(schedule) > 0


def test_edge_case_7_uneven_modules():
    """
    Test Case 7: Careers with very different numbers of modules
    Expected: Should balance workload appropriately
    """
    print("\n" + "="*70)
    print("TEST 7: Uneven Module Distribution")
    print("="*70)
    
    careers = [
        {
            'career': {
                'domain': 'Simple Career',
                'modules': [
                    {'name': 'Task 1', 'duration': 60},
                    {'name': 'Task 2', 'duration': 60}
                ]
            }
        },
        {
            'career': {
                'domain': 'Complex Career',
                'modules': [
                    {'name': f'Task {i}', 'duration': 45}
                    for i in range(1, 11)  # 10 modules
                ]
            }
        }
    ]
    
    module = GoalBasedModule(daily_hours=3.0)
    schedule = module.generate_daily_schedule(careers)
    
    print(f"✓ Schedule generated: {len(schedule)} days")
    
    # Verify all tasks are scheduled
    total_tasks = 2 + 10  # 12 total
    scheduled_tasks = sum(len(day['tasks']) for day in schedule)
    
    assert scheduled_tasks == total_tasks, f"Missing tasks! Expected {total_tasks}, got {scheduled_tasks}"
    print(f"✓ All {total_tasks} tasks scheduled correctly")
    
    module.print_schedule()
    
    return len(schedule) > 0


def test_progress_tracking():
    """
    Test Case 8: Progress tracking functionality
    Expected: Should accurately track completion
    """
    print("\n" + "="*70)
    print("TEST 8: Progress Tracking")
    print("="*70)
    
    careers = [
        {
            'career': {
                'domain': 'Test Career',
                'modules': [
                    {'name': 'Task 1', 'duration': 60},
                    {'name': 'Task 2', 'duration': 60},
                    {'name': 'Task 3', 'duration': 60}
                ]
            }
        }
    ]
    
    module = GoalBasedModule(daily_hours=2.0)
    schedule = module.generate_daily_schedule(careers)
    
    # Initial progress
    progress = module.track_progress()
    assert progress['completed'] == 0, "Initial progress should be 0"
    assert progress['percentage'] == 0, "Initial percentage should be 0"
    print(f"✓ Initial progress: {progress['completed']}/{progress['total']} (0%)")
    
    # Complete first task
    first_task_id = schedule[0]['tasks'][0]['id']
    module.update_progress(first_task_id, True)
    
    progress = module.track_progress()
    assert progress['completed'] == 1, "Should have 1 completed task"
    print(f"✓ After 1 task: {progress['completed']}/{progress['total']} ({progress['percentage']:.1f}%)")
    
    # Complete all tasks
    for day in schedule:
        for task in day['tasks']:
            module.update_progress(task['id'], True)
    
    progress = module.track_progress()
    assert progress['goal_achieved'] == True, "Goal should be achieved"
    print(f"✓ All tasks completed: {progress['completed']}/{progress['total']} (100%)")
    
    return True


def run_all_tests():
    """Run all edge case tests"""
    print("\n" + "="*70)
    print("GOAL-BASED SCHEDULING - COMPREHENSIVE EDGE CASE TESTS")
    print("="*70)
    
    tests = [
        ("High Time Availability", test_edge_case_1_high_availability),
        ("Low Time Availability", test_edge_case_2_low_availability),
        ("Single Career", test_edge_case_3_single_career),
        ("Multiple Careers", test_edge_case_4_many_careers),
        ("Empty Input", test_edge_case_5_empty_input),
        ("Large Task", test_edge_case_6_large_task),
        ("Uneven Modules", test_edge_case_7_uneven_modules),
        ("Progress Tracking", test_progress_tracking)
    ]
    
    passed = 0
    failed = 0
    
    for test_name, test_func in tests:
        try:
            result = test_func()
            if result or result is None:
                passed += 1
                print(f"✅ {test_name}: PASSED")
            else:
                failed += 1
                print(f"❌ {test_name}: FAILED")
        except Exception as e:
            failed += 1
            print(f"❌ {test_name}: ERROR - {str(e)}")
    
    print("\n" + "="*70)
    print(f"TEST RESULTS: {passed} passed, {failed} failed")
    print("="*70)
    
    return failed == 0


if __name__ == "__main__":
    run_all_tests()
```

```python
RUN_FILE = """
Main application runner
Launch CareerBuddy application
"""

# Import system utilities for modifying Python path
import sys
from pathlib import Path

# Add project root to path
# Ensures Python can locate modules inside project directories
sys.path.insert(0, str(Path('careerbuddy').parent))

from models.career_model import CareerBuddyModel
from ui.auth import show_login_page
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML
import json
import os
from datetime import datetime

# Global state
model = CareerBuddyModel()
users_db = {}
output_area = widgets.Output()
USERS_FILE = 'careerbuddy_users.json'

def load_users_db():
    """Load users from file"""
    # Reads JSON file → converts created timestamp string back to datetime → returns dictionary
    if os.path.exists(USERS_FILE):
        with open(USERS_FILE, 'r') as f:
            data = json.load(f)
            for username in data:
                data[username]['created'] = datetime.fromisoformat(data[username]['created'])
            return data
    return {}

def start_application():
    """Start the CareerBuddy application"""
    global users_db
    users_db = load_users_db() # Loads existing users into memory

    # Display the main welcome banner using HTML
    display(HTML("""
    <div style='background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); 
                padding: 40px; border-radius: 15px; text-align: center; color: white; margin: 20px 0;
                box-shadow: 0 10px 40px rgba(0,0,0,0.2);'>
        <h1 style='font-size: 48px; margin: 0 0 10px 0;'>🎯 CareerBuddy AI Agent</h1>
        <h3 style='margin: 10px 0; font-weight: 300;'>Mesa Framework - Organized Structure</h3>
        <p style='margin: 20px 0 0 0; font-size: 16px; opacity: 0.9;'>
            Hybrid Agent System with Persistent Storage
        </p>
    </div>
    """))
    # Clear screen and display login form dynamically inside output area
    with output_area:
        clear_output(wait=True)
        show_login_page(model, users_db, output_area)
    # Render widget container below banner
    display(output_area)

if __name__ == "__main__":
    print()
    start_application() # Entry Point - File executed directly
    
```
